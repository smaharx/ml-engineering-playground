## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What this WILL be able to say:
- "These signals are correlated with pages that decline"
- "This model identifies 80% of pages that ACTUALLY declined (in hindsight)"
- "Pages with high days_since_update AND low impressions are directionally associated with decline"

What this will NEVER be:
- Proof that updating a page CAUSES it to grow back
- Prediction of Google's algorithm
- Proof of causation (only correlation)
- 100% accurate for all pages

Why the caveat?
We're predicting from observation, not from experiments. We see what happened,
but we don't know what would have happened if we'd taken different actions.
The model finds patterns, but those patterns might not be the *cause* of decline.
